## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

The unit of analysis for this dataset is a **single user's search query**. Each row represents a unique search query submitted by a user.

The time window covered by this data spans from **January 1, 2023, to March 31, 2023** (Q1 2023).

```python
import pandas as pd

data = {
    'user_id': [1, 1, 2, 3, 2],
    'query_id': [101, 102, 201, 301, 202],
    'query_text': ['red shoes', 'size 10', 'best restaurants', 'flight tickets', 'coffee maker'],
    'query_timestamp': ['2023-01-05 10:00:00', '2023-01-05 10:05:00', '2023-02-15 14:30:00', '2023-03-20 08:00:00', '2023-03-25 11:15:00']
}
df = pd.DataFrame(data)
df['query_timestamp'] = pd.to_datetime(df['query_timestamp'])

print('Sample of the data (one row = one search query):')
display(df.head())

# --- Verify the Time Window ---
min_date = df['query_timestamp'].min()
max_date = df['query_timestamp'].max()
print(f'\nData time window: from {min_date.date()} to {max_date.date()}')

print(f'\nNumber of unique query IDs: {df["query_id"].nunique()} (should match number of rows if each row is a unique query)')
print(f'Total number of rows: {len(df)}')
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Even with a clean data contract, there are inherent limitations to what this search query data can reveal:

*   **User Intent vs. Query Text:** The `query_text` field captures what a user typed, but not necessarily their underlying intent. A query like "red shoes" could mean "buy red shoes," "red shoes near me," or "history of red shoes." Without further interaction data (e.g., clicks, purchases), we can only infer intent.
*   **Small Sample Size and Time Window:** The current dummy dataset is very small (5 queries over ~3 months). This limits any robust conclusions about user behavior, trends, or seasonality. Real-world data would need a much larger volume and longer time span.
*   **Lack of Contextual Information:** We don't have information about the user's location, device, previous search history (beyond the current query), or the time of day, which could all influence search behavior.
*   **No Negative Signals:** The data only shows *successful* searches (i.e., queries that were submitted). It doesn't capture abandoned searches, misspellings that led to no results, or users who couldn't find what they were looking for. This creates a bias towards what users *did* search for, not what they *failed* to find.
*   **Attribution Challenges:** If a user performs multiple searches for the same item, it's hard to attribute which specific search ultimately led to a conversion (if we had a label for it). `query_id` is unique per query, but doesn't link queries from the same user effectively without additional session/user tracking.
*   **Historical Bias:** If the search platform or underlying product catalog changed significantly before or during the data collection period, the historical data might not accurately reflect current user behavior or product availability.
